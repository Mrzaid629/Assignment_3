### 1. What is the purpose of an if statement?
An if statement is used to check a condition in Python. If the condition is true, the statements inside the if block are executed.

Example:

age = 20

if age >= 18:
    print("Eligible to vote")

### 2. Explain the difference between if, if-else and if-elif-else.



if	- Executes a block only when the condition is true.

if-else	Executes one block if the condition is true and another if it is false.

if-elif-else	Checks multiple conditions and executes the block for the first true condition.







```python
### 3. Write a program to check whether a number is positive, negative or zero.
num = int(input("Enter a number: "))

if num > 0:
    print("Positive number")
elif num < 0:
    print("Negative number")
else:
    print("Zero")
```

```python
### 4. Write a program to check whether a number is even or odd.

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

```python
### 5. Write a program to find the greatest among three numbers.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if a >= b and a >= c:
    print("Greatest number is", a)
elif b >= a and b >= c:
    print("Greatest number is", b)
else:
    print("Greatest number is", c)
```

```python
### 6. Write a program to check whether a given year is a leap year.

year = int(input("Enter a year: "))

if year % 400 == 0:
    print("Leap year")
elif year % 100 == 0:
    print("Not a leap year")
elif year % 4 == 0:
    print("Leap year")
else:
    print("Not a leap year")
```

```python
## 7. Write a program to assign grades based on a student's marks.

marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 60:
    print("Grade C")
elif marks >= 35:
    print("Grade D")
else:
    print("Fail")
```

```python
### 8. Write a program to check whether a person is eligible to vote.

age = int(input("Enter your age: "))

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

```python
### 9. Write a program to calculate an electricity bill using different usage slabs.

units = int(input("Enter electricity units: "))

if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = (100 * 2) + (units - 100) * 3
else:
    bill = (100 * 2) + (100 * 3) + (units - 200) * 5

print("Electricity bill = ₹", bill)
```

### 10. What is a nested if statement? Give a suitable situation where it can be used.
A nested if statement is an if statement written inside another if statement. It is used when a second condition must be checked after the first condition is satisfied.

```python
## Example: Checking voting eligibility and voter registration.

age = int(input("Enter your age: "))
registered = input("Are you registered? (yes/no): ")

if age >= 18:
    if registered == "yes":
        print("You can vote")
    else:
        print("Please register to vote")
else:
    print("You are not eligible to vote")
```
